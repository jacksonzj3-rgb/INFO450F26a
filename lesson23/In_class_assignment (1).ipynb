```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('Advertising_F.csv')
```

```python
df.head(10)
```

```python
px.scatter(df,x='TV',y="radio",
           labels={'TV':'Tv add spend','radio':'Radio add spend'})
```

```python
sales_per_city=df.groupby('City')['sales'].sum().reset_index()
sales_per_city
```

```python
px.bar(sales_per_city,x='City',y="sales",title="data is fun")
```

```python
fig_map = px.scatter_map(
    df,
    lat='latitude',
    lon='longitude',
    size='sales',
    color='sales',
    hover_name='City',
    zoom=3,
    title='Sales Distribution by City Map',
    map_style='carto-positron'
)
fig_map.show()
```

```python
%%writefile my_script.py

import pandas as pd
import streamlit as st
import plotly.express as px

df=pd.read_csv('Advertising_F.csv')
st,title("Advertising Effectiveness Zion")

st.sidebar.head("Filters")
x_var=st.selectbox("X axis",["TV","newspaper"])

#show scatterplot
fig=px.scatter(df,x=x_var,y="sales",
           labels={'TV':'Tv add spend','radio':'Radio add spend'})

st.plotly_chart(fig)

#show the barplot
sales_per_city=df.groupby('City')['sales'].sum().reset_index()
sales_per_city
px.bar(sales_per_city,x='City',y="sales",title="data is fun")

st.plotly_chart(fig)

#show the dataframe
st.dataframe(df)

```
